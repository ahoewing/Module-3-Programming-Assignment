### 7.4 - Create a List

```python

things = ["mozzarella", "cinderella", "salmonella"]
print(things)
```

### 7.5 - Capitalize an Element

```python

things[1].capitalize()
print(things)
```

No, it did not change the element in the list because capitalize() returns a new string without changing the original.

### 7.6 - Make an Element Uppercase

```python

things[0] = things[0].upper()
print(things)
```

### 9.2 - Generator Function

```python

def get_odds():
    for number in range(10):
        if number % 2 != 0:
            yield number

for index, number in enumerate(get_odds(), start=1):
    if index == 3:
        print(number)
        break
```
